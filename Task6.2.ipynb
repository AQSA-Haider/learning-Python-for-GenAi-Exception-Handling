# Bill Calculator with Error Handling
# given a list of all prices:

```python
prices=[120,350,'abc',500,-200,800]
```

# Writing code that :
# iterates over the list
# tries to add onli valid numbers (positive numericals) prices to the 'total'.
# Handles:
# ~TypeError:if value is not a number.
# ~Custom Exceptionusing raise ValueError(Negetive price not allowed)
# prints the running total.
# Expected Behaviour was:
# Skips invalid items but continues processing.

```python
total=0
for i in prices:
    try:
        if not isinstance(i,(int,float)) or isinstance(i,bool):
            raise TypeError(f"Invalid item  '{i }' :  Not a number.")
        if i<0:
            raise ValueError("Negetive prices not allowed")
        total +=i
        print(f"Added {i} , Runningtotal : {total}")

    except TypeError as t:
        print(f"Skipping Item : {t}")
    except ValueError as v:
        print(f"Skipping {i} : {v}")
print(f"\nFinal Total: {total}")
```
