# File Reader with Exception Handling
# Asking the user for a file name: 

```python
file = input("Enter a file name: \n")
```

# Trying to open and read the file
# Handle:
# ~FileNotFoundError
# ~PermissionError
# if successful prints first three line of the file
# Use finally to print:"file operation completed"

```python
try:
    with open(file,'r') as f:
        for i in range(3):
            line=f.readline()
            if not line:
                break
            print(line.strip())

except FileNotFoundError:
    print("Error: You entered a file that does not exist")
except PermissionError:
    print ("Error: Permission not granted")

finally:
    print("File operation attempted")
```
